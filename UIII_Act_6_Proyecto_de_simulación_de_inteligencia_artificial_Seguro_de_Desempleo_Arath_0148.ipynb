# Seguro de desempleo

```python
# Simulador de IA: Asistente de Seguro de Desempleo

print("==========================================================")
print("  BIENVENIDO AL ASISTENTE VIRTUAL DE SEGURO DE DESEMPLEO  ")
print("==========================================================")
print("Hola, soy tu asesor de IA. Te ayudaré a saber si eres elegible")
print("para el subsidio y a calcular tu beneficio estimado.")
print("----------------------------------------------------------\n")

# Variable de control para el ciclo interactivo
continuar_asistencia = "1"

while continuar_asistencia == "1":
    print("--- NUEVA CONSULTA ---")

    # Entrada de datos del usuario
    motivo_despido = input("¿El motivo de tu desempleo fue despido injustificado o fin de contrato? (si/no): ")
    meses_cotizados = input("¿Cuántos meses trabajaste y cotizaste en tu último empleo?: ")
    meses_cotizados = int(meses_cotizados) # Convertimos el texto a número entero

    ultimo_sueldo = input("¿Cuál era tu sueldo mensual bruto en tu último empleo (en USD)?: ")
    ultimo_sueldo = float(ultimo_sueldo) # Convertimos a número con decimales

    print("\n[IA Procesando Datos...]")

    # --- SISTEMA DE DECISIÓN DE LA IA ---

    # Regla 1: El desempleo NO debe ser por renuncia voluntaria (debe ser despido injustificado o fin de contrato)
    # Regla 2: Haber cotizado un mínimo de meses (por ejemplo, mínimo 6 meses)

    if motivo_despido == "si" and meses_cotizados >= 6:
        print("\n¡Resultado: ELEGIBLE!")
        print("Cumples con las condiciones legales para recibir el seguro de desempleo.")

        # Cálculo del beneficio basado en el tiempo cotizado (Simulación de IA experta)
        if meses_cotizados >= 6 and meses_cotizados <= 12:
            porcentaje = 0.60  # Recibe el 60% de su sueldo
            tiempo_ayuda = "4 meses"
        elif meses_cotizados > 12 and meses_cotizados <= 24:
            porcentaje = 0.70  # Recibe el 70% de su sueldo
            tiempo_ayuda = "6 meses"
        else:
            porcentaje = 0.80  # Recibe el 80% de su sueldo por alta antigüedad
            tiempo_ayuda = "12 meses"

        # Calcular el monto mensual
        monto_mensual = ultimo_sueldo * porcentaje

        # Aplicar un tope máximo de ayuda económica (ejemplo: tope de $2000)
        if monto_mensual > 2000:
            monto_mensual = 2000
            print("* Nota: Tu cálculo superaba el límite legal, se aplicó el tope máximo.")

        print(f"- Tu apoyo mensual estimado será de: ${monto_mensual:.2f} USD")
        print(f"- Duración del beneficio: {tiempo_ayuda}")

    elif motivo_despido == "no":
        print("\nResultado: NO ELEGIBLE.")
        print("El seguro de desempleo no aplica si la baja fue por renuncia voluntaria o despido justificado.")

    elif meses_cotizados < 6:
        print("\nResultado: NO ELEGIBLE.")
        print("No cumples con el tiempo mínimo de cotización requerido (Mínimo 6 meses).")

    else:
        print("\nNo se pudo determinar. Verifica que hayas respondido correctamente.")

    # --- CONTROL DEL CICLO INTERACTIVO ---
    print("\n----------------------------------------------------------")
    print("¿Deseas realizar otra consulta de desempleo?")
    print("1. Sí, analizar otro caso.")
    print("2. No, salir del asistente.")
    continuar_asistencia = input("Elige una opción (1 o 2): ")

print("\n==========================================================")
print("  Gracias por usar el Asistente Virtual. ¡Mucho éxito!   ")
print("==========================================================")
```
